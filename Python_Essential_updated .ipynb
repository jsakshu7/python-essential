## List Slicing

```python
# Example 1: Slice first 3 items
letters = ["A","B", "C", "D", "E", "F"]
print("letters[:3]  :",letters[:3]) 

# Example 2: Slice from index 2 to end
print("letters[2:]  :",letters[2:]) 

# Example 3: Slice with step
print("letters[::2]  :",letters[::2])  

# Example 3: Slice with step
print("letters[2:5:2]  :",letters[2:5:2])  
```

## Membership Operator

```python
# Example 1: Using `in`
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)  
```

```python
# Example 2: Using `not in`
print("grape" not in fruits)  
```

```python
# Example 3: Check character in string
print("apple" in fruits)
```

```python
# Example 4: Check characters in string
print("ep" in fruits)
```
